```python
import string as s
import random as r
print ("letters and numbers = ",s.ascii_letters + s.digits)

mixPasswd=r.sample(s.ascii_letters + s.digits, 5)
print (mixPasswd)
mixPasswd="".join(mixPasswd)
print ("mixPasswd1 = ",mixPasswd)

special="#@!~%^&*()_+=-[]{}|"
mixPasswd=r.sample(special + s.ascii_letters + s.digits, 8)
print (mixPasswd)
mixPasswd="".join(mixPasswd)
print ("mixPasswd = ",mixPasswd)
```
