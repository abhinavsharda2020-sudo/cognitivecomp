```python
import numpy as np
a= np.array([10,20,30,40,50])
print(a+2)
print(a*3)
print(a/2)
```

```python
print(a[::-1])
x = np.array([1,2,3,4,5,1,2,1,1,1]) 
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3]) 
values, counts= np.unique(x, return_counts=True)
mfx=values[np.argmax(counts)]
ix=np.where(x==mfx)
print("most frequent in x:",mfx)
print("indices in x:",ix)

values, counts= np.unique(y, return_counts=True)
mfy=values[np.argmax(counts)]
iy=np.where(y==mfy)
print("most frequent in y:",mfy)
print("indices in y:",iy)
```

```python
abhinav= np.linspace(10, 100, 25)
print("Array:", abhinav)
print("Dimensions:", abhinav.ndim)
print("Shape:", abhinav.shape)
print("Total elements:", abhinav.size)
print("Data type:", abhinav.dtype)
print("Total bytes:", abhinav.nbytes)
transposed = abhinav.reshape(25, 1)
print("Transposed using reshape:", transposed)
print("Using .T attribute (no change on 1-D array):", abhinav.T)
```

```python
ucs420_abhinav = np.array([[10, 20, 30, 40],
                          [50, 60, 70, 80],
                          [90, 15, 20, 35]])
print("Array:", ucs420_abhinav)
print("Mean:", np.mean(ucs420_abhinav))
print("Median:", np.median(ucs420_abhinav))
print("Max:", np.max(ucs420_abhinav))
print("Min:", np.min(ucs420_abhinav))
print("Unique elements:", np.unique(ucs420_abhinav))
reshaped_ucs420_abhinav = ucs420_abhinav.reshape(4, 3)
print("Reshaped (4x3):\n", reshaped_ucs420_abhinav)
resized_ucs420_abhinav = np.resize(ucs420_abhinav, (2, 3))
print("Resized (2x3):\n", resized_ucs420_abhinav)
```
