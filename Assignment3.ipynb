```python
dat={
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR','IT','IT','Marketing','Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10','2010-11-25'],
    'Gender': ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]}
d=pd.DataFrame(dat)
print(d)
```

```python
a= d['Salary'].mean()
t= d['Bonus'].sum()
y= d['Age'].min()
h= d['Rating'].max()

print("Average Salary:", a)
print("Total Bonus:", t)
print("Youngest Employee's Age:", y)
print("Highest Rating:", h)
```

```python
d_sorted = d.sort_values(by='Salary', ascending=False)
print(d_sorted)
```

```python
def categorize(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'
d['Performance_Category'] = d['Rating'].apply(categorize)
print(d)
```

```python
print(d.isnull().sum())
print("\n     ")
d = d.rename(columns={'Employee_ID': 'ID'})
print(d.columns)
print("\n     ")
m= d[d['Years_of_Experience'] > 5]
it= d[d['Department'] == 'IT']
print(m)
print(it)
```

```python
d['Tax'] = d['Salary'] * 0.10
print(d)
```
