```python
import pandas as pd
df= pd.DataFrame({'number': [1024170306]})
dig = df['number'].astype(str).str.split('', expand=True).iloc[:, 1:]
print(dig)
roll_no= "1024170306"
fixed_entries = [ {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.",
                   "keywords": "fee cost price charge", "category": "billing"},
                  {"question": "how to reset password", "answer": "Go to Settings > Reset Password.",
                   "keywords": "password reset login", "category": "account"},
                  {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",
                   "keywords": "hours timing open time", "category": "general"},
                  {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.",
                   "keywords": "pay payment upi fee", "category": "billing"}, ] 

lsd= [int(d) for d in roll_no[-2:]]
print(lsd)
category=["billing", "account", "general"]
```

```python
p=[]
for d in lsd:
    cat=category[d%3]
    if cat=="billing":
        entry= {"question": "how do i get a refund for an overcharged payment", 'answer': "Submit a request from the billing section",
                "keywords": "refund overcharge billing payment","category":cat}
    elif cat=="account":
        entry={"question": "how do i update my mobile number", "answer": "Go to Profile > Contact Details and update your mobile number, then verify via OTP.",
               "keywords": "mobile number update account", "category": cat}
    else:
        entry= {"question": "how do i contact customer support", "answer": "You can reach customer support via the Help Center chat",
                "keywords": "contact support help query", "category": cat}
    p.append(entry)
faq_df=pd.DataFrame(fixed_entries+p)
print(faq_df)
```

```python
def scoreq(query, df):
    query_words= set(query.lower().split())
    scores=[]
    for i in range(len(df)):
        k= df["keywords"][i].lower().split()
        q= df["question"][i].lower().split()
        s=0
        for w in query_words:
            if w in k:
                s=s+1
            if w in q:
                s=s+1
        scores.append(s)
    df["score"]= scores
    result=df[df["score"] >0]
    result= result.sort_values("score", ascending=False)
    return result
print(scoreq,"how do i pay fee",faq_df)
```

```python
def same_category(category_name, df):
    result = df[df["category"] == category_name]
    return result[["question", "category"]]

demo_category = p[0]["category"]
print(same_category(demo_category, faq_df))
```

```python
def score_query_ties(query, df):
    query_words = query.lower().split()
    scores = []
    for i in range(len(df)):
        kw = df["keywords"][i].lower().split()
        q = df["question"][i].lower().split()
        s = 0
        for w in query_words:
            if w in kw:
                s = s + 1
            if w in q:
                s = s + 1
        scores.append(s)
    df["score"] = scores
    result = df[df["score"] > 0]
    result = result.sort_values("score", ascending=False)

    if len(result) == 0:
        print("No match for:", query)
        return result

    top = result["score"].max()
    ties = result[result["score"] == top]

    if len(ties) > 1:
        print("Tie for query:", query)
        print(ties)
    else:
        print("Best match for:", query)
        print(ties)
    return result

score_query_ties("fee", faq_df)
score_query_ties("reset password login", faq_df)
```
