```python
import random
random.seed(1024170306)
numbers = [random.randint(100, 900) for _ in range(100)]

od=[x for x in numbers if x%2 != 0]
print(od, len(od))
ev=[x for x in numbers if x%2 == 0]
print(ev, len(ev))

pn= [x for x in numbers if x > 1 and all(x % i != 0 for i in range(2, int(x**0.5) + 1))]
print("the prime numbers are: ",pn)

m= max(set(numbers), key=numbers.count)
c= numbers.count(m)
print("the number: ",m,"the count: ",c)
```

```python
roll_no=1024170306
digits=[int(d) for d in str(roll_no)]
A={digit*7 for digit in digits}
B={digit*9 for digit in digits}
print(A)
print(B)
print("Union: ", A.union(B))
print("intersection: ", A.intersection(B))
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))                               #difference gives  value in one set but not in other  
print("Symmetric difference:", A.symmetric_difference(B))      #whereas symmetric difference returns value that are in either set but not in both
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
X = int(input("Enter a value X: "))
A.discard(X)
print("Set A after discarding X:", A)                #it doesnt give error if value isnt present
```

```python
my_dict = {
    "name": "Abhinav",
    "roll_no": "1024170306",
    "branch": "COPC",
    "age": 20,
    "city": "Chandigarh"
}
my_dict["location"]= my_dict.pop("city")
my_dict["CGPA"]= 7.09
my_dict["age"] +=1
print(my_dict)

pop= my_dict.copy()                                # pop() removes the key AND returns its value, while del removes the key 
                                                   # but does not return its value
ddel= my_dict.copy()
del ddel["branch"]
print("using pop: ",pop)
print("using del: ",ddel)

print("Key-value pairs:")
for key, value in my_dict.items():
    print(key, "→", value)

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key does not exist.")

f_dict = {
    "name": "nik",
    "roll_no": "1024170771",
    "branch": "COPC",
    "age": 20,
    "city": "Hyderabad"
}
m_dict = {**my_dict, **f_dict}
print("Merged dictionary:", m_dict)                      # because friend_dict is unpacked second.

s = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("String values only:", s)
```
